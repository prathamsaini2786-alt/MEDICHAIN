```python
import random
from datetime import datetime, timedelta

random.seed(42)
np.random.seed(42)

medicines = {
    "Paracetamol": 500,
    "Amoxicillin": 250,
    "Insulin": 100,
    "Azithromycin": 200,
    "Ibuprofen": 300
}

locations = [
    "Delhi",
    "Chandigarh",
    "Amritsar",
    "Ludhiana",
    "Patiala"
]

suppliers = [
    "Supplier_A",
    "Supplier_B",
    "Supplier_C"
]

data = []

start_date = datetime(2025, 1, 1)

for i in range(365):

    date = start_date + timedelta(days=i)

    for medicine, base_demand in medicines.items():

        for location in locations:

            # Weekly pattern
            day_of_week = date.weekday()

            if day_of_week >= 5:
                weekly_factor = 0.85
            else:
                weekly_factor = 1.0

            # Seasonal pattern
            month = date.month

            if month in [11, 12, 1, 2]:
                seasonal_factor = 1.15
            elif month in [6, 7, 8]:
                seasonal_factor = 0.95
            else:
                seasonal_factor = 1.0

            # Random variation
            noise = np.random.normal(0, 30)

            demand = (
                base_demand
                * weekly_factor
                * seasonal_factor
                + noise
            )

            demand = max(20, int(demand))

            # Inventory
            inventory = random.randint(
                demand * 3,
                demand * 10
            )

            # Supplier
            supplier = random.choice(suppliers)

            # Normal delivery time
            delivery_time = random.randint(2, 7)

            data.append([
                date,
                medicine,
                location,
                demand,
                inventory,
                supplier,
                delivery_time
            ])

real_df = pd.DataFrame(data, columns=[
    "Date",
    "Medicine",
    "Location",
    "Demand",
    "Inventory",
    "Supplier",
    "Delivery_Time"
])

real_df.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Features
features = [
    "Inventory",
    "Delivery_Time",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "IsWeekend",
    "Previous_Demand",
    "7_Day_Average"
]

X_train = train_df[features]
y_train = train_df["Demand"]

X_test = test_df[features]
y_test = test_df["Demand"]

# Identify categorical and numerical features
categorical_features = X_train.select_dtypes(include=["object"]).columns.tolist()
numerical_features = X_train.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", "passthrough", numerical_features)
    ]
)

# Random Forest model
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\nModel Performance")
print("-----------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Features
features = [
    "Medicine",
    "Location",
    "Supplier",
    "Inventory",
    "Delivery_Time",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "IsWeekend",
    "Previous_Demand",
    "7_Day_Average"
]

# Make sure numerical columns are actually numeric
numeric_features = [
    "Inventory",
    "Delivery_Time",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "IsWeekend",
    "Previous_Demand",
    "7_Day_Average"
]

categorical_features = [
    "Medicine",
    "Location",
    "Supplier"
]

for col in numeric_features:
    train_df[col] = pd.to_numeric(train_df[col], errors="coerce")
    test_df[col] = pd.to_numeric(test_df[col], errors="coerce")

# Remove rows with missing values in required features
train_clean = train_df.dropna(subset=features + ["Demand"]).copy()
test_clean = test_df.dropna(subset=features + ["Demand"]).copy()

X_train = train_clean[features]
y_train = train_clean["Demand"]

X_test = test_clean[features]
y_test = test_clean["Demand"]

print("Categorical features:", categorical_features)
print("Numerical features:", numeric_features)

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", "passthrough", numeric_features)
    ]
)

# Random Forest
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\nModel Performance")
print("-----------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
# Recalculate lag features
real_df["Previous_Demand"] = (
    real_df
    .groupby(["Medicine", "Location"])["Demand"]
    .shift(1)
)

real_df["7_Day_Average"] = (
    real_df
    .groupby(["Medicine", "Location"])["Demand"]
    .transform(
        lambda x: x.shift(1).rolling(7).mean()
    )
)

# Remove rows where historical features don't exist
real_df = real_df.dropna(
    subset=["Previous_Demand", "7_Day_Average"]
).copy()

# Recreate train/test split
cutoff_date = pd.Timestamp("2025-10-21")

train_df = real_df[real_df["Date"] < cutoff_date].copy()
test_df = real_df[real_df["Date"] >= cutoff_date].copy()

print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))

print("\nMissing historical features:")
print(
    train_df[["Previous_Demand", "7_Day_Average"]]
    .isna()
    .sum()
)
```

```python
# Define features and target

features = [
    "Medicine",
    "Location",
    "Supplier",
    "Inventory",
    "Delivery_Time",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "IsWeekend",
    "Previous_Demand",
    "7_Day_Average"
]

X_train = train_df[features]
y_train = train_df["Demand"]

X_test = test_df[features]
y_test = test_df["Demand"]

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("\nTraining missing values:")
print(X_train.isna().sum().sum())

print("\nTesting missing values:")
print(X_test.isna().sum().sum())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Categorical and numerical features
categorical_features = [
    "Medicine",
    "Location",
    "Supplier"
]

numerical_features = [
    "Inventory",
    "Delivery_Time",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "IsWeekend",
    "Previous_Demand",
    "7_Day_Average"
]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            "passthrough",
            numerical_features
        )
    ]
)

# Random Forest model
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MODEL PERFORMANCE")
print("-----------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```
